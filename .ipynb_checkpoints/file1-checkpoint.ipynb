```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.2f}'.format)
```

```python
BASE = "/Users/dhruvjani/2026projects/micromobility-safety-deprivation-uk/dataset/"

collisions = pd.read_csv(BASE + "dft-road-casualty-statistics-collision-last-5-years.csv", low_memory=False)
casualties = pd.read_csv(BASE + "dft-road-casualty-statistics-casualty-last-5-years.csv", low_memory=False)
vehicles   = pd.read_csv(BASE + "dft-road-casualty-statistics-vehicle-last-5-years.csv",  low_memory=False)

print(f"Collisions : {len(collisions):,} rows, {collisions.shape[1]} columns")
print(f"Casualties : {len(casualties):,} rows, {casualties.shape[1]} columns")
print(f"Vehicles   : {len(vehicles):,} rows, {vehicles.shape[1]} columns")
```

```python
# ── COLLISIONS ──────────────────────────────────────────────────────────────

print("COLLISIONS DATASET")
print(f"Rows: {len(collisions):,}")
print(f"Columns: {collisions.shape[1]}")
print("\nFirst 5 Rows")
display(collisions.head())
print("\nData Types")
print(collisions.dtypes)
print("\nMissing Values")
missing_col = collisions.isnull().sum()
print(missing_col[missing_col > 0])
print("\nBasic Statistics")
display(collisions.describe())
print("\nYears Present")
print(collisions['collision_year'].value_counts().sort_index())


# ── CASUALTIES ──────────────────────────────────────────────────────────────

print("\n\nCASUALTIES DATASET")
print(f"Rows: {len(casualties):,}")
print(f"Columns: {casualties.shape[1]}")
print("\nFirst 5 Rows")
display(casualties.head())
print("\nData Types")
print(casualties.dtypes)
print("\nMissing Values")
missing_cas = casualties.isnull().sum()
print(missing_cas[missing_cas > 0])
print("\nBasic Statistics")
display(casualties.describe())
print("\nCasualty Type Breakdown")
print(casualties['casualty_type'].value_counts().sort_index())


# ── VEHICLES ────────────────────────────────────────────────────────────────

print("\n\nVEHICLES DATASET")
print(f"Rows: {len(vehicles):,}")
print(f"Columns: {vehicles.shape[1]}")
print("\nFirst 5 Rows")
display(vehicles.head())
print("\nData Types")
print(vehicles.dtypes)
print("\nMissing Values")
missing_veh = vehicles.isnull().sum()
print(missing_veh[missing_veh > 0])
print("\nBasic Statistics")
display(vehicles.describe())
print("\nVehicle Type Breakdown")
print(vehicles['vehicle_type'].value_counts().sort_index())
```

```python
print("COLLISIONS")
print(collisions.dtypes)
print("\nCASUALTIES")
print(casualties.dtypes)
```

```python
ped_cyc = casualties[casualties['casualty_type'].isin([0, 1])].copy()
ped_cyc['casualty_type_label'] = ped_cyc['casualty_type'].map({0: 'Pedestrian', 1: 'Cyclist'})

print(f"Pedestrian and cyclist casualties: {len(ped_cyc):,}")
print(ped_cyc['casualty_type_label'].value_counts())
```

```python
print("Years in dataset:")
print(collisions['collision_year'].value_counts().sort_index())
```

```python
cols_needed = [
    'collision_index', 'collision_year', 'collision_severity',
    'date', 'longitude', 'latitude',
    'urban_or_rural_area', 'local_authority_ons_district',
    'speed_limit', 'light_conditions', 'weather_conditions'
]

cols_needed = [c for c in cols_needed if c in collisions.columns]

merged = ped_cyc.merge(collisions[cols_needed], on='collision_index', how='left')
merged['severity_label'] = merged['collision_severity'].map({1: 'Fatal', 2: 'Serious', 3: 'Slight'})
merged['area_label']     = merged['urban_or_rural_area'].map({1: 'Urban', 2: 'Rural'})
merged = merged.dropna(subset=['longitude', 'latitude'])

print(f"Merged rows: {len(merged):,}")
print(merged['severity_label'].value_counts())
```

```python
cols_needed = [
    'collision_index', 'collision_year', 'collision_severity',
    'date', 'longitude', 'latitude',
    'urban_or_rural_area', 'local_authority_ons_district',
    'speed_limit', 'light_conditions', 'weather_conditions'
]

cols_needed = [c for c in cols_needed if c in collisions.columns]
print("Columns found:", cols_needed)  # confirm collision_year is in there

merged = ped_cyc.merge(collisions[cols_needed], on='collision_index', how='left')
merged['severity_label'] = merged['collision_severity'].map({1: 'Fatal', 2: 'Serious', 3: 'Slight'})
merged['area_label']     = merged['urban_or_rural_area'].map({1: 'Urban', 2: 'Rural'})
merged = merged.dropna(subset=['longitude', 'latitude'])

print(f"Merged rows: {len(merged):,}")
print(merged['severity_label'].value_counts())
```

```python
print(merged.columns.tolist())
```

```python
urban = merged[merged['urban_or_rural_area'] == 1].copy()

# Use collision_year_y (from collisions table) or whichever exists
year_col = 'collision_year_y' if 'collision_year_y' in urban.columns else 'collision_year'

print(f"Urban pedestrian and cyclist casualties: {len(urban):,}")
print("\nBy year:")
print(urban[year_col].value_counts().sort_index())
print("\nBy casualty type:")
print(urban['casualty_type_label'].value_counts())
```

```python
ped_cyc = casualties[casualties['casualty_type'].isin([0, 1])].copy()
ped_cyc['casualty_type_label'] = ped_cyc['casualty_type'].map({0: 'Pedestrian', 1: 'Cyclist'})

print(f"Total pedestrian and cyclist casualties: {len(ped_cyc):,}")
print(ped_cyc['casualty_type_label'].value_counts())
```

```python
collision_cols = [
    'collision_index', 'collision_year', 'collision_severity',
    'date', 'longitude', 'latitude',
    'urban_or_rural_area', 'local_authority_ons_district',
    'speed_limit', 'light_conditions', 'weather_conditions',
    'lsoa_of_accident_location'
]

collisions_slim = collisions[collision_cols].copy()

print(f"Collisions slim shape: {collisions_slim.shape}")
print(collisions_slim.head())
```

```python
escooter = vehicles.groupby('collision_index')['escooter_flag'].max().reset_index()

print(f"Escooter table shape: {escooter.shape}")
print(f"Collisions involving e-scooter: {escooter['escooter_flag'].sum():,}")
print(escooter['escooter_flag'].value_counts())
```

```python
master = ped_cyc.merge(collisions_slim, on='collision_index', how='left')
master = master.merge(escooter, on='collision_index', how='left')

print(f"Master shape after merge: {master.shape}")
print(master.head())
```

```python
master = master[master['urban_or_rural_area'] == 1].copy()

print(f"Urban only rows: {len(master):,}")
print(master['urban_or_rural_area'].value_counts())
```

```python
master.replace(-1, np.nan, inplace=True)
master = master.dropna(subset=['lsoa_of_accident_location', 'longitude', 'latitude'])

print(f"Rows after cleaning: {len(master):,}")
print(f"Rows dropped: {141989 - len(master):,}")
```

```python
# Drop the duplicate year column from casualties side
master = master.drop(columns=['collision_year_x'])
master = master.rename(columns={'collision_year_y': 'collision_year'})

# Add readable labels
master['severity_label'] = master['collision_severity'].map({1: 'Fatal', 2: 'Serious', 3: 'Slight'})

print(f"Final master shape: {master.shape}")
print(f"\nBy year:")
print(master['collision_year'].value_counts().sort_index())
print(f"\nSeverity:")
print(master['severity_label'].value_counts())
```

```python
import os

os.makedirs("data/processed", exist_ok=True)

master.to_csv("data/processed/master_ped_cyc_urban_2020_2024.csv", index=False)

print("Saved successfully.")
print("Location: data/processed/master_ped_cyc_urban_2020_2024.csv")
print(f"Rows: {len(master):,}")
print(f"Columns: {master.shape[1]}")
```
